Question 1: What is a random variable in probability theory?

A random variable in probability theory is a variable that assigns a numerical value to the outcome of a random experiment.

It is a function that maps outcomes from the sample space (S) of a random process to real numbers (ℝ).

The value it takes is not fixed in advance but depends on the outcome of the experiment.

Types of Random Variables:

Discrete Random Variable

Takes countable values (e.g., number of heads in 3 coin tosses → {0,1,2,3}).

Question 2: What are the types of random variables?

In probability theory, random variables are mainly classified into two types:

1. Discrete Random Variable

Takes countable values (finite or infinite).

Examples:

Number of heads in 3 coin tosses → {0, 1, 2, 3}.

Number of students absent in a class → 0, 1, 2, …

👉 Think of it as values you can list or count.

2. Continuous Random Variable

Takes uncountably infinite values within an interval.

Examples:

Height of students (can be 150.2 cm, 150.25 cm, etc.).

Time taken to finish a race.

👉 Think of it as values that come from measurement, not counting.

Quick Analogy for Students:

Discrete → Counting chocolates in a box 🍫 (0,1,2,…).

Continuous → Measuring milk in a bottle 🥛 (1.2 L, 1.25 L, …).

Question 3: Explain the difference between discrete and continuous distributions.

1. Discrete Distribution

Based on a discrete random variable (countable outcomes).

Probability is given for specific values.

Represented using a Probability Mass Function (PMF).

Example:

Tossing a coin 3 times, random variable = number of heads.

Possible values: {0, 1, 2, 3}.

Each has its own probability.

👉 Think: "Exact counts" → 2 heads, 3 absent students, 5 chocolates.

2. Continuous Distribution

Based on a continuous random variable (uncountably infinite outcomes).

Probability is given for an interval (not for a single value).

Represented using a Probability Density Function (PDF).

Example:

Measuring height of students in cm.

You can’t say probability(Height = 160.5) = some value.

Instead, you say probability(160 ≤ Height ≤ 170).

👉 Think: "Measurements" → time, height, weight, distance.

Key Differences Table
Feature	Discrete Distribution	Continuous Distribution
Variable type	Discrete (countable)	Continuous (measurable)
Probability	Assigned to exact values	Assigned to intervals
Function used	PMF (Probability Mass Function)	PDF (Probability Density Function)
Example	Number of heads in coin toss	Height of students

✨ Shortcut to remember:

Discrete = D = Digit/Counting

Continuous = C = Continuous measurement

Question 9: Write a Python program to generate 1000 random numbers from a normal
distribution with mean = 50 and standard deviation = 5. Compute its mean and standard
deviation using NumPy, and draw a histogram to visualize the distribution.
(Include your Python code and output in the code box below.)


Here’s the Python solution with code, calculation, and visualization:

```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate 1000 random numbers from Normal(μ=50, σ=5)
data = np.random.normal(loc=50, scale=5, size=1000)

# Step 2: Compute mean and standard deviation
mean_val = np.mean(data)
std_val = np.std(data)

print("Sample Mean:", mean_val)
print("Sample Standard Deviation:", std_val)

# Step 3: Plot histogram
plt.hist(data, bins=30, density=True, color='skyblue', edgecolor='black')
plt.title("Histogram of Normal Distribution (mean=50, sd=5)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

✅ Expected Output (varies slightly each run because of randomness):
Sample Mean: ~50.1
Sample Standard Deviation: ~5.0
The histogram will show a bell-shaped curve, centered near 50, with spread around 5.

Step 2: Python Code

```python
import numpy as np
import scipy.stats as stats

# Daily sales data (sample)
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Convert to NumPy array
data = np.array(daily_sales)

# Sample statistics
mean_sales = np.mean(data)
std_sales = np.std(data, ddof=1)   # sample standard deviation
n = len(data)

# 95% Confidence Interval using CLT
confidence = 0.95
alpha = 1 - confidence
z_value = stats.norm.ppf(1 - alpha/2)  # Z = 1.96 approx

margin_of_error = z_value * (std_sales / np.sqrt(n))
lower_bound = mean_sales - margin_of_error
upper_bound = mean_sales + margin_of_error

print("Sample Mean Sales:", mean_sales)
print("95% Confidence Interval: [{:.2f}, {:.2f}]".format(lower_bound, upper_bound))
```

Step 3: Expected Output (values may vary slightly due to sample variation)
Sample Mean Sales: 248.25
95% Confidence Interval: [241.93, 254.57]


✅ Interpretation:
We are 95% confident that the true average daily sales for the company (over 2 years) lies between ≈ 242 and 255 units.
